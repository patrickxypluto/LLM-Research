```python
# Load the structured aviation manual
from google.colab import drive
drive.mount('/content/drive')
file_path = "/content/drive/MyDrive/Haystack/Dataset/manual.json"
import json
with open(file_path, "r", encoding="utf-8") as file:
  manual=json.load(file)

contexts = [context['context'] for context in manual]
```

```python
# functions that will be used for generating a prompt
def prompt_for_generating_QAs(chain_type):
  if chain_type == "LDS":
    return prompt_for_generating_QAs_LDS
  elif chain_type == "ESS":
    return prompt_for_generating_QAs_ESS
  else:
    return prompt_for_generating_QAs_RNS

def generate_template_from_one_id_chain(chain, chain_type, contexts):
  template = prompt_for_generating_QAs(chain_type)
  for i in range(len(chain)):
    template += f"Context {i+1}\n"
    template += f"{contexts[chain[i]]}\n"
  template += "-------------------------------------------------------------------\n"
  return template
```

```python
# Use gemini to generate QAs
import google.generativeai as genai
from google.colab import userdata
GOOGLE_API_KEY = API_KEY
genai.configure(api_key=GOOGLE_API_KEY)
gemini_model = genai.GenerativeModel('models/gemini-2.5-flash')
```

```python
# Load the manual_context_ids_chains
file_path = "/content/drive/MyDrive/Haystack/Dataset/manual_context_ids_chains.json"
import json
with open(file_path, "r", encoding="utf-8") as file:
  manual_context_ids_chains=json.load(file)

LDS_chains = manual_context_ids_chains["LDS"]
ESS_chains = manual_context_ids_chains["ESS"]
RNS_chains = manual_context_ids_chains["RNS"]
```

```python
# function for converting the response of gemini model into the wanted format
def convert_QAs_to_lists(QAs, QAs_ids):
  qas_lst = QAs.split('\n')
  qs_lst = []
  as_lst = []
  ids_lst = []
  for line in qas_lst:
    if line.startswith("QAs"):
      continue
    elif line.startswith("Context"):
      line = line.strip("Context")
      line = line.strip()
      id_index = int(line)-1
      id = QAs_ids[id_index]
    elif line.startswith("Q"):
      line = line[5:]
      line = line.strip()
      qs_lst.append(line)
      ids_lst.append(id)
    elif line.startswith("A"):
      line = line[5:]
      line = line.strip()
      as_lst.append(line)
  return qs_lst, as_lst, ids_lst
```
